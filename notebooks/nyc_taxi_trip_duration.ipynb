<!-- Introduction -->
<p style="font-size: 16px; line-height: 1.7; text-align: justify;">
    Afin d’assurer une <strong>exécution stable, performante et adaptée à un environnement Windows</strong>, 
    la configuration d’<strong>Apache Spark</strong> constitue une étape essentielle du projet.
    La configuration de Spark commence par la mise en place des <strong>paramètres réseau et d’exécution de base</strong>, 
    nécessaires pour éviter les erreurs de communication et optimiser le traitement de 
    <strong>données volumineuses</strong>.
</p>

<p style="font-size: 16px; line-height: 1.7;">
    Nous avons commencé par les étapes suivantes :
</p>

<!-- ===================== STEP 1 ===================== -->
<p style="font-size: 17px; margin-top: 18px;">
    <strong style="color:#1565c0;">1. Forçage de l’adresse IP locale (127.0.0.1)</strong>
</p>

<p style="font-size: 16px; line-height: 1.7; text-align: justify;">
    Sous Windows, Apache Spark peut rencontrer des problèmes de liaison réseau, notamment lors de 
    l’initialisation du <strong>Spark Driver</strong>.  
    Afin d’éviter les erreurs de type <em>Connection refused</em> ou <em>BindException</em>, 
    l’adresse IP locale a été explicitement définie.
</p>

<ul style="font-size: 15px; line-height: 1.7;">
    <li>Garantit que Spark écoute uniquement sur <strong>l’interface locale</strong></li>
    <li>Améliore la <strong>stabilité</strong> lors de l’exécution en mode local</li>
    <li>Évite les <strong>conflits réseau</strong> avec d’autres interfaces 
        (VPN, Wi-Fi, Ethernet)</li>
</ul>

<!-- ===================== STEP 2 ===================== -->
<p style="font-size: 17px; margin-top: 20px;">
    <strong style="color:#1565c0;">2. Suppression des SparkContext zombies</strong>
</p>

<p style="font-size: 16px; line-height: 1.7; text-align: justify;">
    Avant la création d’une nouvelle session Spark, il est indispensable de fermer toute session existante,
    en particulier dans un environnement <strong>Jupyter Notebook</strong> où les cellules peuvent être exécutées
    plusieurs fois.
</p>

<ul style="font-size: 15px; line-height: 1.7;">
    <li>Empêche la création de <strong>plusieurs SparkContext simultanés</strong></li>
    <li>Évite les <strong>conflits de ports</strong></li>
    <li>Libère correctement la <strong>mémoire</strong> et les <strong>ressources système</strong></li>
</ul>

<!-- ===================== STEP 3 ===================== -->
<p style="font-size: 17px; margin-top: 20px;">
    <strong style="color:#1565c0;">
        3. Activation de l’optimisation adaptative 
        (Adaptive Query Execution – AQE)
    </strong>
</p>

<p style="font-size: 16px; line-height: 1.7; text-align: justify;">
    Pour améliorer les performances sur des datasets contenant 
    <strong>des millions de lignes</strong>, l’optimisation adaptative de 
    <strong>Spark SQL</strong> a été activée.
</p>

<p style="font-size: 15px; margin-left: 10px;">
    <code style="color:#2e7d32;">spark.sql.adaptive.enabled = true</code><br>
    <code style="color:#2e7d32;">spark.sql.adaptive.coalescePartitions.enabled = true</code>
</p>

<p style="font-size: 16px; line-height: 1.7;">
    Ces options permettent à Spark :
</p>

<ul style="font-size: 15px; line-height: 1.7;">
    <li>D’<strong>optimiser dynamiquement</strong> le plan d’exécution</li>
    <li>De réduire le nombre de <strong>partitions inutiles</strong></li>
    <li>D’améliorer l’utilisation de la <strong>mémoire</strong> et du <strong>CPU</strong></li>
    <li>D’accélérer les <strong>traitements analytiques</strong> et 
        les <strong>opérations de Machine Learning</strong></li>
</ul>



```python


import os

# Forcer Spark à rester sur l’IP locale (important sous Windows)
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"
os.environ["PYSPARK_SUBMIT_ARGS"] = (
    "--conf spark.driver.bindAddress=127.0.0.1 pyspark-shell"
)

# Nettoyer les anciennes sessions Spark (si elles existent)
try:
    from pyspark.sql import SparkSession
    old_spark = SparkSession.builder.getOrCreate()
    if old_spark:
        print("Ancienne session Spark détectée → arrêt.")
        old_spark.stop()
except:
    pass

# Forcer la suppression d’un SparkContext zombie
try:
    from pyspark import SparkContext
    SparkContext._active_spark_context = None
except:
    pass

print("Environnement Spark nettoyé. Prêt à créer une nouvelle session.")

import findspark
findspark.init()

from pyspark.sql import SparkSession

# Création d’une nouvelle session Spark
spark = (
    SparkSession.builder.appName("NYC Taxi Duration Prediction")
    .master("local[*]")  # utilise tous les cœurs
    .config("spark.driver.bindAddress", "127.0.0.1")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.sql.adaptive.enabled", "true")
    .config("spark.sql.adaptive.coalescePartitions.enabled", "true")
    .getOrCreate()
)

print(f"Spark {spark.version} démarré avec succès ✔️")
```

<h5>Visualisation des trajets sur une carte</h5>

```python
import folium
#carte centree sur newyork
carte=folium.Map(location=[40.75, -73.97], zoom_start=11)

# Échantillon de 1000 points pour visualisation rapide
sample = pandas_df.sample(10000, random_state=42)

# Ajouter les points
for _, row in sample.iterrows():
    # Pickup
    folium.CircleMarker(
        location=[row['pickup_latitude'], row['pickup_longitude']],
        radius=1,
        color='blue',
        fill=True,
        fill_opacity=0.5
    ).add_to(carte)
    
    # Dropoff
    folium.CircleMarker(
        location=[row['dropoff_latitude'], row['dropoff_longitude']],
        radius=1,
        color='green',
        fill=True,
        fill_opacity=0.5
    ).add_to(carte)
carte
```

```python
# Filtrer les points aberrants
outliers = pandas_df[
    (pandas_df['pickup_latitude'] < 40) | 
    (pandas_df['pickup_latitude'] > 41) |
    (pandas_df['pickup_longitude'] < -75) | 
    (pandas_df['pickup_longitude'] > -72)
]

# Créer une autre carte pour eux
map_outliers = folium.Map(location=[40.75, -73.97], zoom_start=4)  # zoom plus large

for _, row in outliers.sample(98,replace=True).iterrows():
    folium.CircleMarker(
        location=[row['pickup_latitude'], row['pickup_longitude']],
        radius=3,
        color='red',
        fill=True,
        fill_opacity=0.6
    ).add_to(map_outliers)
map_outliers
```

<h1>Clean DataSets</h1>
<h3 style="color:#2c3e50;">4. Nettoyage des données (Data Cleaning)</h3> <p style="font-size:16px; line-height:1.7; text-align:justify;"> Le nettoyage des données est une étape centrale du pipeline de data science. C’est ici que nous préparons le dataset pour la modélisation en le rendant <strong>propre, cohérent et réaliste</strong>. Les données brutes contiennent souvent des anomalies, doublons, valeurs aberrantes ou informations incohérentes qui peuvent fausser les modèles. </p> <h4 style="color:#34495e;">Règles appliquées :</h4> <ul style="font-size:15px; line-height:1.6;"> <li><strong>Suppression des doublons :</strong> élimination des lignes répétées pour éviter de biaiser les analyses.</li> <li><strong>Filtrage du nombre de passagers :</strong> seuls les trajets avec 1 à 6 passagers sont conservés, afin d’exclure les entrées irréalistes.</li> <li><strong>Suppression des trajets sans déplacement :</strong> trajets où les coordonnées de départ et d’arrivée sont identiques.</li> <li><strong>Filtrage géographique :</strong> conservation uniquement des trajets situés à New York City (coordonnées plausibles).</li> <li><strong>Filtrage des durées aberrantes :</strong> suppression des trajets de moins de 60 secondes et de plus de 2 heures.</li> </ul> <h4 style="color:#34495e;">Résultats :</h4> <p style="font-size:16px; line-height:1.7; text-align:justify;"> Après ces opérations, le dataset est devenu <strong>propre, réaliste et cohérent</strong>. Les statistiques montrent une réduction significative des valeurs aberrantes et un filtrage des trajets non pertinents. Environ <strong>98.89%</strong> des données initiales ont été conservées, ce qui constitue un compromis entre qualité et volume de données. </p> <p style="font-size:16px; line-height:1.7; text-align:justify;"> Une visualisation des points de départ après nettoyage confirme que les trajets sont bien concentrés dans la zone de NYC, et que les anomalies géographiques ont été éliminées. Ce dataset est désormais prêt pour l’étape suivante : l’ingénierie des features et la modélisation prédictive. </p>

```python
# Faire une copie
clean_df = train_df.alias("clean_df")

# 1. Supprimer doublons
initial_count = clean_df.count()
clean_df = clean_df.dropDuplicates()
final_count = clean_df.count()
print(f"Doublons supprimés: {initial_count - final_count}")

# 2. Garder passagers réalistes (1 à 6)
clean_df = clean_df.filter(
    (col("passenger_count") >= 1) &
    (col("passenger_count") <= 6)
)
print(f"Après filtrage passagers 1-6: {clean_df.count()}")

# 3. Supprimer trajets pickup = dropoff
clean_df = clean_df.filter(
    (col("pickup_latitude") != col("dropoff_latitude")) | 
    (col("pickup_longitude") != col("dropoff_longitude"))
)
print(f"Après suppression trajets identiques: {clean_df.count()}")

# 4. Filtrage géographique NYC
clean_df = clean_df.filter(
    (col("pickup_longitude").between(-74.3, -73.6)) &
    (col("pickup_latitude").between(40.5, 40.9)) &
    (col("dropoff_longitude").between(-74.3, -73.6)) &
    (col("dropoff_latitude").between(40.5, 40.9))
)

print(f"Après filtrage géographique: {clean_df.count()}")

# 5. Filtrer durées aberrantes
clean_df = clean_df.filter(
    (col("trip_duration").between(60, 2*3600))  
)
print(f"Après filtrage durée: {clean_df.count()}")

# Pourcentage conservé
print(f"Pourcentage données conservées: {(clean_df.count()/train_df.count()*100):.2f}%")
```

<h1>Calcul de distance Haversine</h1>

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
import math

# === 1. CALCUL DE LA DISTANCE AVEC FONCTIONS SPARK NATIVES ===
# Au lieu d'UDF, on utilise des expressions Spark SQL directes
clean_df = clean_df.withColumn(
    "distance_km",
    expr("""
        6371 * 2 * asin(sqrt(
            pow(sin(radians(dropoff_latitude - pickup_latitude) / 2), 2) +
            cos(radians(pickup_latitude)) * 
            cos(radians(dropoff_latitude)) * 
            pow(sin(radians(dropoff_longitude - pickup_longitude) / 2), 2)
        ))
    """)
)

# === 2. FEATURES TEMPORELLES AVANCÉES (DÉJÀ BON) ===
clean_df = clean_df.withColumn("pickup_hour", hour("pickup_datetime"))\
                   .withColumn("pickup_day", dayofweek("pickup_datetime"))\
                   .withColumn("pickup_month", month("pickup_datetime"))\
                   .withColumn("is_weekend", 
                              when((col("pickup_day") == 1) | (col("pickup_day") == 7), 1).otherwise(0))\
                   .withColumn("is_rush_hour",
                              when(((col("pickup_hour") >= 7) & (col("pickup_hour") <= 10)) | 
                                   ((col("pickup_hour") >= 16) & (col("pickup_hour") <= 19)), 1).otherwise(0))

# === 3. FEATURES GÉOGRAPHIQUES SANS UDF ===
# Définir les coordonnées des aéroports
jfk_lat, jfk_lon = 40.6413, -73.7781
lga_lat, lga_lon = 40.7769, -73.8740

# Fonction de distance pour les aéroports (en degrés approximatifs - plus simple)
def airport_distance_expression(airport_lat, airport_lon, point_lat_col, point_lon_col, radius_deg=0.03):
    """Crée une expression SQL pour vérifier la proximité d'un aéroport"""
    return f"""
        CASE 
            WHEN (POW({point_lat_col} - {airport_lat}, 2) + POW({point_lon_col} - {airport_lon}, 2)) < POW({radius_deg}, 2)
            THEN 1 
            ELSE 0 
        END
    """

# Appliquer les features aéroports avec des expressions SQL
clean_df = clean_df.withColumn(
    "pickup_near_jfk", 
    expr(airport_distance_expression(jfk_lat, jfk_lon, "pickup_latitude", "pickup_longitude"))
).withColumn(
    "dropoff_near_jfk", 
    expr(airport_distance_expression(jfk_lat, jfk_lon, "dropoff_latitude", "dropoff_longitude"))
).withColumn(
    "pickup_near_lga", 
    expr(airport_distance_expression(lga_lat, lga_lon, "pickup_latitude", "pickup_longitude"))
).withColumn(
    "dropoff_near_lga", 
    expr(airport_distance_expression(lga_lat, lga_lon, "dropoff_latitude", "dropoff_longitude"))
)
clean_df = clean_df.fillna({'distance_km': 0})

# === 4. VITESSE MOYENNE ===
clean_df = clean_df.withColumn("speed_kmh", 
                              (col("distance_km") / (col("trip_duration") / 3600)))

# === 5. FILTRER LES VITESSES IRRÉALISTES ===
clean_df = clean_df.filter(
    (col("speed_kmh").isNotNull()) &
    (col("speed_kmh") > 0) &
    (col("speed_kmh") < 120)
)


# Vérification des nouvelles features
print("=== NOUVELLES FEATURES CRÉÉES ===")
clean_df.select("distance_km", "pickup_hour", "is_weekend", "is_rush_hour", 
               "pickup_near_jfk", "speed_kmh", "trip_duration").show(10)

print(f"Taille finale après feature engineering: {clean_df.count()} lignes")
```

<h3 style="color:#2c3e50;">7. Entraînement des Modèles de Régression</h3> <p style="font-size:16px; line-height:1.7; text-align:justify;"> Après le nettoyage des données et la création des features, l’étape suivante consiste à entraîner différents modèles de régression pour prédire la durée des trajets à New York. La variable cible utilisée est <code>log_duration</code>, la transformation logarithmique de <code>trip_duration</code>, afin de stabiliser la variance et réduire l’influence des valeurs extrêmes. </p> <h4 style="color:#34495e;">7.1 Modèles entraînés</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Trois types de modèles de régression ont été entraînés sur les données normalisées (<code>scaledFeatures</code>) : <ul> <li><strong>Régression Linéaire :</strong> modèle simple servant de baseline, avec une régularisation pour limiter l’overfitting.</li> <li><strong>Random Forest :</strong> modèle d’ensemble basé sur des arbres de décision, capable de capturer des relations non linéaires entre les features et la durée.</li> <li><strong>Gradient Boosted Trees (GBT) :</strong> modèle d’ensemble séquentiel, plus puissant que la Random Forest, qui corrige progressivement les erreurs des arbres précédents.</li> </ul> Ces modèles ont été entraînés uniquement sur le jeu d’entraînement pour éviter toute fuite d’information vers le jeu de validation. </p> <h4 style="color:#34495e;">7.2 Évaluation avec conversion logarithmique</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Étant donné que la variable cible a été transformée en logarithme, les prédictions des modèles sont d’abord obtenues sous forme logarithmique. Une conversion inverse (<code>exp(prediction)</code>) est appliquée pour obtenir la durée en secondes. Trois métriques principales sont utilisées pour évaluer les modèles : <ul> <li><strong>RMSE (Root Mean Squared Error) :</strong> écart quadratique moyen, sensible aux grandes erreurs.</li> <li><strong>MAE (Mean Absolute Error) :</strong> erreur absolue moyenne, plus robuste aux outliers.</li> <li><strong>R² :</strong> coefficient de détermination, indiquant la proportion de variance expliquée par le modèle.</li> </ul> Cette approche permet de comparer directement les prédictions en secondes avec la vérité terrain, tout en conservant les avantages de la transformation logarithmique. </p> <h4 style="color:#34495e;">7.3 Résultats et comparaison des modèles</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Après l’évaluation sur le jeu de validation, chaque modèle est comparé selon ses métriques. Par exemple : <ul> <li>La <strong>Régression Linéaire</strong> sert de référence et permet d’identifier les limites des modèles linéaires pour ce type de problème.</li> <li>Le <strong>Random Forest</strong> capture mieux les relations complexes entre les features et la durée, réduisant souvent l’erreur moyenne.</li> <li>Le <strong>GBT</strong> tend à fournir les meilleures performances en terme de R² et RMSE grâce à l’optimisation séquentielle des arbres.</li> </ul> Un tableau synthétique des performances permet de visualiser rapidement les différences entre modèles et de sélectionner le modèle le plus approprié pour la prédiction de la durée des trajets. </p> <h4 style="color:#34495e;">7.4 Visualisation et contrôle des prédictions</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Des exemples de prédictions sont affichés pour vérifier la cohérence avec les valeurs réelles. Cette étape qualitative complète l’évaluation quantitative et assure que les modèles génèrent des prédictions plausibles pour des trajets réalistes. </p>

```python
from pyspark.sql.functions import exp, col
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.ml.regression import LinearRegression, RandomForestRegressor, GBTRegressor

# === ENTRAÎNEMENT DES MODÈLES (AVEC LOG) ===

# 1. RÉGRESSION LINÉAIRE
lr = LinearRegression(
    featuresCol="scaledFeatures", 
    labelCol="log_duration",  # ← CHANGÉ
    maxIter=10,
    regParam=0.3
)
lr_model = lr.fit(train_scaled)

# 2. RANDOM FOREST
rf = RandomForestRegressor(
    featuresCol="scaledFeatures",
    labelCol="log_duration",  # ← CHANGÉ
    numTrees=20,
    maxDepth=10,
    seed=42
)
rf_model = rf.fit(train_scaled)

# 3. GRADIENT BOOSTING TREES
gbt = GBTRegressor(
    featuresCol="scaledFeatures",
    labelCol="log_duration",  # ← CHANGÉ
    maxIter=20,
    maxDepth=6,
    seed=42
)
gbt_model = gbt.fit(train_scaled)

# === NOUVELLE FONCTION D'ÉVALUATION (AVEC CONVERSION) ===
def afficher_resultats_avec_log(nom_modele, predictions):
    """
    Évalue un modèle qui prédit log_duration.
    Convertit d'abord les prédictions en secondes.
    """
    # Étape CRUCIALE : convertir la prédiction (log) en secondes
    predictions = predictions.withColumn("prediction_seconds", exp(col("prediction")))
    
    # Créer les évaluateurs pour comparer avec trip_duration
    rmse_evaluator = RegressionEvaluator(
        labelCol="trip_duration",           # Vérité en secondes
        predictionCol="prediction_seconds", # Prédiction convertie en secondes
        metricName="rmse"
    )
    
    mae_evaluator = RegressionEvaluator(
        labelCol="trip_duration",
        predictionCol="prediction_seconds",
        metricName="mae"
    )
    
    r2_evaluator = RegressionEvaluator(
        labelCol="trip_duration",
        predictionCol="prediction_seconds",
        metricName="r2"
    )
    
    # Calculer les métriques
    rmse = rmse_evaluator.evaluate(predictions)
    mae = mae_evaluator.evaluate(predictions)
    r2 = r2_evaluator.evaluate(predictions)
    
    print(f"\n=== {nom_modele} ===")
    print(f"RMSE : {rmse:.2f} sec   ({rmse/60:.2f} minutes)")
    print(f"MAE  : {mae:.2f} sec   ({mae/60:.2f} minutes)")
    print(f"R²   : {r2:.4f}")
    
    # Afficher quelques exemples
    print("\nExemples de prédictions (5 premières lignes):")
    predictions.select("trip_duration", "log_duration", "prediction", "prediction_seconds").show(5)
    
    return predictions

# === ÉVALUATION DES MODÈLES ===

print("=" * 60)
print("ÉVALUATION AVEC TRANSFORMATION LOGARITHMIQUE")
print("=" * 60)

# 1. Régression linéaire
lr_predictions = lr_model.transform(val_scaled)
lr_predictions_clean = afficher_resultats_avec_log("Linear Regression (log)", lr_predictions)

# 2. Random Forest
rf_predictions = rf_model.transform(val_scaled)
rf_predictions_clean = afficher_resultats_avec_log("Random Forest (log)", rf_predictions)

# 3. Gradient Boosted Trees
gbt_predictions = gbt_model.transform(val_scaled)
gbt_predictions_clean = afficher_resultats_avec_log("Gradient Boosted Trees (log)", gbt_predictions)

# === COMPARAISON DES MODÈLES ===
print("\n" + "=" * 60)
print("SYNTHÈSE DES PERFORMANCES")
print("=" * 60)

# Créer un DataFrame de comparaison
comparison_data = {
    "Modèle": ["Linear Regression", "Random Forest", "GBT"],
    "R²": [],
    "RMSE (minutes)": [],
    "MAE (minutes)": []
}

# Fonction pour extraire les métriques
def extraire_metriques(predictions, modele_nom):
    predictions = predictions.withColumn("prediction_seconds", exp(col("prediction")))
    
    rmse = RegressionEvaluator(
        labelCol="trip_duration",
        predictionCol="prediction_seconds",
        metricName="rmse"
    ).evaluate(predictions)
    
    mae = RegressionEvaluator(
        labelCol="trip_duration",
        predictionCol="prediction_seconds",
        metricName="mae"
    ).evaluate(predictions)
    
    r2 = RegressionEvaluator(
        labelCol="trip_duration",
        predictionCol="prediction_seconds",
        metricName="r2"
    ).evaluate(predictions)
    
    return r2, rmse/60, mae/60

# Remplir le tableau de comparaison
for modele_nom, predictions in [
    ("Linear Regression", lr_predictions),
    ("Random Forest", rf_predictions),
    ("GBT", gbt_predictions)
]:
    r2, rmse_min, mae_min = extraire_metriques(predictions, modele_nom)
    print(f"{modele_nom:25} R²={r2:.4f}  RMSE={rmse_min:.2f} min  MAE={mae_min:.2f} min")
```

<h3 style="color:#2c3e50;">8. Prédiction sur le Jeu de Test et Évaluation Finale</h3> <p style="font-size:16px; line-height:1.7; text-align:justify;"> Une fois le modèle final sélectionné (<strong>Gradient Boosted Trees</strong>), il est appliqué au jeu de test et au jeu de validation pour générer des prédictions de durée de trajet. Cette étape consiste à préparer le jeu de test de manière identique au jeu d’entraînement, en reproduisant toutes les transformations et le feature engineering. </p> <h4 style="color:#34495e;">8.1 Préparation du jeu de test</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Le jeu de test est transformé en ajoutant toutes les features calculées pour l’entraînement : <ul> <li><strong>Distance haversine :</strong> distance à vol d’oiseau entre le point de prise en charge et le point de dépose.</li> <li><strong>Distance Manhattan :</strong> approximation urbaine de la distance sur les axes N/S et E/O.</li> <li><strong>Direction du trajet :</strong> angle du trajet en degrés.</li> <li><strong>Features temporelles :</strong> heure, jour, mois, indicateurs week-end et heures de pointe.</li> <li><strong>Proximité des aéroports :</strong> indicateurs binaires pour JFK et LGA, au départ et à l’arrivée.</li> <li><strong>Encodage de store_and_fwd_flag :</strong> conversion en valeur binaire (1 pour Y, 0 pour N).</li> </ul> Toutes ces transformations garantissent que le modèle reçoit les mêmes types de features que lors de l’entraînement. </p> <h4 style="color:#34495e;">8.2 Application du pipeline de transformations</h4> <p style="font-size:15px; line-height:1.6; text-align:justify;"> Le pipeline de transformations (<code>VectorAssembler</code> et <code>StandardScaler</code>) utilisé pour l’entraînement est appliqué au jeu de test.
